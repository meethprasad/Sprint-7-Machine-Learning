## 1. What is Unsupervised Learning?

Unsupervised Learning is a type of Machine Learning where the model learns patterns, structures, or relationships from data **without labelled target values**.

In supervised learning, we have:

**Input Features → Target/Label → Prediction**

In unsupervised learning, we have:

**Input Features → Find Hidden Patterns/Structure**

The algorithm tries to discover useful patterns in the data by itself.

### Example

Suppose a shopping website has customer information such as:
- Age
- Annual Income
- Spending Score

If we do not know which customers belong to which group, we can use clustering to automatically find groups such as:
- Low-income customers
- High-income frequent spenders
- High-income low spenders

No group labels are provided beforehand.

## 2. When is Unsupervised Learning Used?

Unsupervised Learning is especially useful when **labelled data is unavailable**.

Creating labels manually can be expensive, slow, or difficult. Instead of manually labelling every data point, unsupervised algorithms can discover patterns automatically.

### Examples

- Customer groups are unknown → use clustering
- Millions of documents have no categories → discover document groups
- Large datasets contain many features → use dimensionality reduction
- Unknown patterns exist in transaction data → identify unusual structures

Therefore:

**No target/label available → Unsupervised Learning can be useful.**

## 3. Supervised vs Unsupervised Learning

| Feature | Supervised Learning | Unsupervised Learning |
|---|---|---|
| Labels | Required | Not required |
| Target variable | Available | Usually unavailable |
| Main goal | Predict target | Discover patterns/structure |
| Example | Spam detection | Customer segmentation |
| Common algorithms | Linear Regression, Logistic Regression, Decision Tree | K-Means, DBSCAN, PCA |

### Simple Difference

**Supervised:** Learn from known answers.

**Unsupervised:** Find patterns when answers are not provided.

## 5. Clustering

Clustering is an unsupervised learning technique that divides data into groups called **clusters**.

Data points within the same cluster should be relatively similar, while data points in different clusters should be relatively different.

### Example

A company can use customer age and spending score to automatically identify different customer groups.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Create sample unlabeled data
X, _ = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.0,
    random_state=42
)

print("Shape of data:", X.shape)
print("First 5 rows:\n", X[:5])
```

## 6. K-Means Clustering

K-Means is one of the most commonly used clustering algorithms.

### Basic Steps

1. Choose the number of clusters K.
2. Initialize cluster centroids.
3. Assign each data point to the nearest centroid.
4. Recalculate the centroids.
5. Repeat the assignment and update steps.
6. Stop when the clusters become stable.

K-Means requires us to specify the number of clusters, K.

```python
# Create and train K-Means model
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

print("Cluster labels for first 20 points:")
print(labels[:20])

print("\nCluster centers:")
print(kmeans.cluster_centers_)
```

```python
# Visualize the clusters
plt.figure(figsize=(8, 6))
    
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    s=40
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200,
    label='Centroids'
)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

## 7. Choosing the Number of Clusters

One common method for selecting K is the **Elbow Method**.

The Elbow Method calculates the model's inertia for different values of K.

Inertia measures how close data points are to their assigned cluster centroid.

We look for an approximate 'elbow' where increasing K further gives smaller improvements.

```python
inertias = []
    
for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    inertias.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertias, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(range(1, 11))
plt.show()
```

## 8. Dimensionality Reduction

Dimensionality reduction reduces the number of features in a dataset while retaining as much useful information as possible.

### Why reduce dimensions?

- Makes data easier to visualize
- Reduces computational complexity
- Can remove redundant information
- Can help with high-dimensional datasets
- Makes some datasets easier to analyze

### PCA

PCA stands for **Principal Component Analysis**.

PCA transforms the original features into a smaller number of new features called **principal components**.

The first principal component captures the greatest amount of variance possible, followed by the second, and so on.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load Iris dataset
iris = load_iris()
X_iris = iris.data

print("Original shape:", X_iris.shape)

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_iris)

# Reduce 4 features to 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Reduced shape:", X_pca.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())
```

```python
# Visualize PCA result
plt.figure(figsize=(8, 6))
    
plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    s=40
)

plt.title("PCA: 4 Dimensions Reduced to 2")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```
