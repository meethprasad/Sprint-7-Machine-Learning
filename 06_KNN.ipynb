## 1. What is KNN?

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm.

KNN makes predictions by looking at the nearest training data points.

For classification, KNN uses majority voting.

For regression, KNN usually takes the average value of the nearest neighbors.

## 2. Distance

Distance tells KNN how close two data points are.

A smaller distance means the points are more similar or closer.

A larger distance means the points are farther apart.

## 3. Euclidean Distance

Euclidean distance is the straight-line distance between two points.

For two points:

A = (x1, y1)
B = (x2, y2)

Formula:

d = sqrt((x2 - x1)^2 + (y2 - y1)^2)

```python
import numpy as np

point_a = np.array([2, 3])
point_b = np.array([5, 7])

distance = np.sqrt(np.sum((point_a - point_b) ** 2))

print("Point A:", point_a)
print("Point B:", point_b)
print("Euclidean Distance:", round(distance, 4))
```

## 4. Neighbors

Neighbors are the training data points closest to a new data point.

The value of K determines how many neighbors are considered.

Example:
- K = 1 → one nearest neighbor
- K = 3 → three nearest neighbors
- K = 5 → five nearest neighbors

## 5. Classification

In KNN classification, the class is selected using majority voting among the nearest neighbors.

Example with K = 5:

3 neighbors → Class 1
2 neighbors → Class 0

Prediction → Class 1

## 6. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 7. Load Dataset

We will use the Breast Cancer dataset from Scikit-learn.

Target classes:
- 0 = Malignant
- 1 = Benign

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="Target")

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Target classes:", np.unique(y))
```

## 8. Inspect Dataset

```python
print("First 5 rows:")
print(X.head())

print("\nClass distribution:")
print(y.value_counts())
```

## 9. Train/Test Split

Training data is used to build the model.

Testing data is used to evaluate the model on unseen data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 10. Feature Scaling

KNN is distance-based, so feature scaling is very important.

If one feature has much larger numerical values than another feature, it can dominate the distance calculation.

StandardScaler transforms the features so that they are on a comparable scale.

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## 11. Implement KNN Classification

We will start with K = 5.

The model will look at the 5 nearest training points and use majority voting.

```python
knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train_scaled, y_train)

y_pred = knn.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)

print("K:", 5)
print("Accuracy:", round(accuracy, 4))
```

## 12. Predictions

```python
print("First 10 predicted values:")
print(y_pred[:10])

print("\nFirst 10 actual values:")
print(y_test.iloc[:10].values)
```

## 13. Classification Report

```python
print(classification_report(y_test, y_pred))
```

## 14. Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

## 15. Choosing K

Choosing K is important for KNN performance.

Small K:
- Sensitive to individual data points.
- Can cause overfitting.

Large K:
- Considers many neighbors.
- Can make the model too simple.
- Can cause underfitting.

We can experiment with different K values and compare their accuracy.

```python
k_values = range(1, 21)
accuracies = []

for k in k_values:
    knn_model = KNeighborsClassifier(n_neighbors=k)
    knn_model.fit(X_train_scaled, y_train)
    predictions = knn_model.predict(X_test_scaled)
    accuracy = accuracy_score(y_test, predictions)
    accuracies.append(accuracy)

results = pd.DataFrame({
    "K": list(k_values),
    "Accuracy": accuracies
})

print(results)
```

## 16. Find Best K

```python
best_index = np.argmax(accuracies)
best_k = list(k_values)[best_index]
best_accuracy = accuracies[best_index]

print("Best K:", best_k)
print("Best Accuracy:", round(best_accuracy, 4))
```

## 17. Plot K vs Accuracy

```python
plt.figure(figsize=(8, 5))
plt.plot(list(k_values), accuracies, marker="o")
plt.xlabel("K")
plt.ylabel("Accuracy")
plt.title("KNN: K Value vs Accuracy")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

## 18. Train Final KNN Model

We use the K value that produced the highest test accuracy in our experiment.

```python
final_knn = KNeighborsClassifier(n_neighbors=best_k)

final_knn.fit(X_train_scaled, y_train)

final_predictions = final_knn.predict(X_test_scaled)

final_accuracy = accuracy_score(
    y_test,
    final_predictions
)

print("Final K:", best_k)
print("Final Accuracy:", round(final_accuracy, 4))
```

## 19. KNN Regression

KNN can also be used for regression.

Classification:
- Neighbors vote for a class.

Regression:
- Neighbor values are averaged to produce a numerical prediction.

Example:
Predicting house price using nearby houses.

```python
from sklearn.neighbors import KNeighborsRegressor

X_reg = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6],
    [7],
    [8]
])

y_reg = np.array([
    10,
    20,
    30,
    40,
    50,
    60,
    70,
    80
])

knn_regressor = KNeighborsRegressor(n_neighbors=3)

knn_regressor.fit(X_reg, y_reg)

new_value = np.array([[4.5]])

regression_prediction = knn_regressor.predict(new_value)

print("Input:", new_value[0][0])
print("Predicted value:", round(regression_prediction[0], 2))
```

## 20. Advantages of KNN

1. Simple and easy to understand.
2. Easy to implement.
3. Can be used for classification and regression.
4. No complicated training process.
5. Can model non-linear relationships.
6. Works well with smaller datasets.

## 21. Limitations of KNN

1. Prediction can be slow for large datasets.
2. Requires feature scaling for distance-based calculations.
3. Sensitive to irrelevant features.
4. Choosing the wrong K can reduce performance.
5. Requires storing the training dataset.
6. Can perform poorly with very high-dimensional data.

## 22. Feature Scaling Example

Suppose we have:

Age = 20 to 60
Income = 20,000 to 2,00,000

Income has much larger numerical values.

Without scaling, income can have a much larger influence on distance.

Scaling puts the features on comparable scales.
